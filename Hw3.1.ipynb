```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

## Функція гіпотези реграсії

```python
def hypothesis_function(X, w):
    y_pred = np.dot(X, w)
    return y_pred   
```

## Функція витрат

```python
def loss_function(X, y, w):
    return np.square(hypothesis_function(X, w) - y).sum() / (2 * X.shape[0])
```

## Градієнтний спус

```python
def gradient_step(X, y, w, learning_rate):
    gradient = (X.T @ (hypothesis_function(X, w) - y)) / len(y)
    w -= learning_rate * gradient
    return w
```

## Найкращі параметри w

```python
df = pd.read_csv("Housing.csv")

features = np.array(df[["area", "bedrooms", "bathrooms"]])
y = np.array(df["price"])

def gradient(X, y, learning_rate, num_iter, eps):
    ones = np.ones((df.shape[0], 1))
    X = np.concatenate((ones, features), axis=1)
    w = np.zeros(X.shape[1])
    loss = loss_function(X, y, w)
    loss_history = [loss]
    for _ in range(num_iter):
        w = gradient_step(X, y, w, learning_rate)
        loss = loss_function(X, y, w)
        if abs(loss - loss_history[-1]) < eps:
            loss_history.append(loss)
            break
        loss_history.append(loss)
    return w, loss_history
```

## Аналітичне рішення

```python
def analytical_solution(X, y):
    solution = np.linalg.inv(np.dot(X.T, X)) @ X.T @ y
    return solution
analytical = analytical_solution(X, y)
print(analytical)    
```
