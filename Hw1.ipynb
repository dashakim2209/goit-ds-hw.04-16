# Завдання 1. Непарні числа = -1

# Завдання 3. Вертикальні масиви

```python
arr2a = np.array([[1, 2],
                 [3, 4]
                 ])
arr2b = np.array([[5, 6],
                 [7, 8]
                 ])
print(np.vstack((arr2a, arr2b)))
```

# Завдання 4. Конкатенація 

```python
a = np.array([1,2,3])
part1 = np.repeat(a, 3)
part2 = np.tile(a, 3)
print(np.concatenate((part1, part2)))
```

# Завдання 5. Спільні елементи

```python
a1 = np.array([1,2,3,2,3,4,3,4,5,6]) 
b1 = np.array([7,2,10,2,7,4,9,4,9,8])
print(np.intersect1d(a1, b1))
```

# Завдання 6. 5 максимальних значень

# Завдання 7. Вилучення NpNan

# Завдання 8. Евклідова відстань

```python
a4 = np.array([1, 2, 3, 4, 5])
b2 = np.array([4, 5, 6, 7, 8])
distance = np.sqrt(np.sum((a4 - b2) ** 2))
print(distance)
```
